```sql
CREATE CATALOG IF NOT EXISTS CASESTUDY;

CREATE SCHEMA IF NOT EXISTS BRIGHTLEARN;


--MATCH THE STORE WITH NAMES WITH THEIR REVENUE

SELECT a.store_id,
        a.store_name,
        store_format,
        SUM(basket_value_zar) AS revenue_per_store
FROM casestudy.brightlearn.stores AS a
INNER JOIN casestudy.brightlearn.transactions AS b
ON a.store_id = b.store_id
GROUP BY a.store_id, a.store_name, store_format;

--FINDING STORE TOP 3 BY REVENUE
SELECT a.store_id,
        a.store_name,
        store_format,
        SUM(basket_value_zar) AS revenue_per_store
FROM casestudy.brightlearn.stores AS a
INNER JOIN casestudy.brightlearn.transactions AS b
ON a.store_id = b.store_id
GROUP BY a.store_id, a.store_name, store_format
ORDER BY revenue_per_store DESC
LIMIT 3;

--MONTH REVEIW DATA
SELECT *
FROM casestudy.brightlearn.transactions;

SELECT MONTHNAME(transaction_date) AS month_,
        SUM(basket_value_zar),
        COUNT(num_items)
FROM casestudy.brightlearn.transactions
GROUP BY month_;


```
