```sql
--CREATE CATALOG
CREATE CATALOG IF NOT EXISTS brightlearn;

--CREATE SCHEMA
CREATE SCHEMA IF NOT EXISTS brightlearn.default




--Exercise 2
--Question 1: List all distinct departments in the students table.

SELECT DISTINCT department
FROM brightlearn.students.exercise2

--Question 2: Get the average age of students per department.
SELECT department, AVG(age) AS avg_age
FROM brightlearn.students.exercise2
GROUP BY department

--Question 3: Show departments with more than 1 student.

SELECT department, COUNT(*) AS student_count
FROM brightlearn.students.exercise2
GROUP BY department
HAVING student_count > 1

--Question 4: Get all students whose age is between 21 and 23
SELECT *
FROM brightlearn.students.exercise2
WHERE age BETWEEN 21 AND 23

--Question 5: `List all students in the IT or HR department who are older than 21`
SELECT *
FROM brightlearn.students.exercise2
WHERE department = 'IT' or department = 'HR' and age > 21

--Question 6: Show total credits per department, only for departments with more than 5 total credits.
SELECT department, SUM(credits) AS total_credits
FROM brightlearn.students.courses
GROUP BY department
HAVING total_credits > 5

--Question 7: List all courses that do not have 4 credits
SELECT *
FROM brightlearn.students.courses
WHERE credits != 4

--Question 8:List all courses that do not have 4 credits
SELECT course_id, course_name, credits
FROM brightlearn.students.courses
SORT BY credits DESC
LIMIT 3

--Question 9: Get the maximum, minimum, and average grade across all enrollments.

SELECT MAX(grade) as max_grade, 
     MIN(grade) as min_grade,
     AVG(grade) as avg_grade
FROM brightlearn.students.enrollments

--Question 10: Count how many enrollments exist per course.

SELECT course_id, COUNT(course_id) as enrollment_count
FROM brightlearn.students.enrollments
GROUP BY course_id

--Question 11: Find total salary and total bonus per department

Select department, SUM(salary) as total_salary, SUM(bonus) as total_bonus
FROM brightlearn.default.salaries
GROUP BY department

--Question 12: Show departments where average salary is above 55,000.
SELECT department, AVG(salary) as avg_salary
FROM brightlearn.default.salaries
GROUP BY department
HAVING avg_salary > 55000

--Question 13: List employees whose salary plus bonus is greater than 60,000
SELECT employee_id, 
          name, 
          department, 
          salary, 
          bonus, 
          salary + bonus as total_compensation
FROM brightlearn.default.salaries
WHERE salary + bonus > 60000

DROP TABLE brightlearn.default.salaries

CREATE TABLE brightlearn.default.salaries (
  employee_id INT,
  name STRING,
  department STRING,
  salary INT,
  bonus INT
);

INSERT INTO brightlearn.default.salaries VALUES
(1, 'Tom', 'IT', 60000, 5000),
(2, 'Jerry', 'HR', 55000, 4000),
(3, 'Spike', 'Finance', 70000, 6000),
(4, 'Tyke', 'IT', 62000, 5500),
(5, 'Butch', 'HR', 54000, 3500)

```
