```sql

--Question 1: Check Who is registered

SELECT * 
FROM brightlearn.default.students
WHERE registered = TRUE

--Question 2: Who is not registered?

SELECT *
FROM brightlearn.default.students
WHERE registered = false

--Questino 3 : Who is registered and is a female?

SELECT *
FROM brightlearn.default.students
Where gender = 'Female' and registered = 'true'

--Question 4: Who is not registered and is a female?
SELECT *
FROM brightlearn.default.students
WHERE registered = FALSE AND gender = 'Female'

--Question 5: Who is registered and is a male?

SELECT *
FROM brightlearn.default.students
WHERE registered = TRUE and gender = 'Male'

--Question 6: Who is not registered and is a male?

SELECT *
FROM brightlearn.default.students
WHERE registered = FALSE and gender = 'Male'

--Question 7: Who is registered and is a female?

SELECT *
FROM brightlearn.default.students
WHERE registered = TRUE and gender = 'Female'

--Question 8: who is registered male or female
SELECT *
FROM brightlearn.default.students
WHERE registered = TRUE

--Question 9:  Who is doing BIO105, female and over 21 years

SELECT *
FROM brightlearn.default.students
WHERE course = 'BIO105' AND gender = 'Female' AND age > 21

--Question 10: Who is doing BIO105 and is 30 or less
SELECT *
FROM brightlearn.default.students
WHERE course = 'BIO105' AND age <= 30 and registered = TRUE

--Question 11: How many male
SELECT COUNT(gender)
FROM brightlearn.default.students
WHERE gender = 'Male'

--Question 12: How many students are doing BIO105?

SELECT COUNT(course)
FROM brightlearn.default.students
WHERE course = "BIO105"
```
