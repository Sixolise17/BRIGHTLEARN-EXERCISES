```sql
--Question 1: Write query using INNER JOIN to return the name of the user and their corresponding amount for all matching orders

SELECT users.name, 
orders_1.amount 
FROM brightlearn.default.users
INNER JOIN brightlearn.default.orders_1 ON users.user_id = orders_1.user_id

--Question 2: Write query using LEFT JOIN to return all users and their order amounts, keeping users who have not placed any orders.

SELECT users.name, 
orders_1.amount 
FROM brightlearn.default.users
LEFT JOIN brightlearn.default.orders_1 ON users.user_id = orders_1.user_id

--Question 3: Write a query using LEFT JOIN to find all users from the users table who have never placed an order.

SELECT users.name,
orders_1.amount
FROM brightlearn.default.users
LEFT JOIN brightlearn.default.orders_1 ON users.user_id = orders_1.user_id
WHERE orders_1.amount IS NULL

--Question 4: Write a query to find the total sum of amount spent by each user using INNER JOIN and GROUP BY
SELECT users.name,
SUM(orders_1.amount) as total_amount
FROM brightlearn.default.users
INNER JOIN brightlearn.default.orders_1 ON users.user_id = orders_1.user_id
GROUP BY users.name
--Question 5: Write a query using LEFT JOIN and COALESCE (or IFNULL) to display $0 or 0 for users who have no orders instead of NULL.
```
