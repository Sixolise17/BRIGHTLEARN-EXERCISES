```sql
--CREATE CATALOG
CREATE  CATALOG IF NOT EXISTS brightlearn;
--CREATE SCHEMA
CREATE SCHEMA IF NOT EXISTS brightlearn.default;
--CREATE TABLE
CREATE TABLE IF NOT EXISTS brightlearn.default.products(
    product_id INT,
    product_name STRING,
    price DECIMAL
);

DROP TABLE brightlearn.default.products;
--INSTERT INTO TABLE
INSERT INTO brightlearn.default.products VALUES
(1, 'Laptop', 1200.00),
(2, 'Phone', 800.00),
(3, 'Keyboard', 45.00),
(4, 'Monitor', 300.00),
(5, 'Mouse', 25.00);

--Question 1: Classify each product by price into the three tiers below

SELECT product_id,
       product_name,
       price,
       CASE
       WHEN price > 1000 THEN 'Expensive'
       WHEN price BETWEEN 100 and 1000 THEN 'Mid_range'
       WHEN price < 100 THEN 'Budget'
       END AS price_category
FROM brightlearn.default.products;


CREATE TABLE IF NOT EXISTS brightlearn.default.orders(
    order_id INT,
    customer_name STRING,
    amount DECIMAL
);

INSERT INTO brightlearn.default.orders VALUES
(1, 'Alice', 105.00),
(2, 'Bob', 560.00),
(3, 'Charlie', 999.99),
(4, 'Diana', 45.50),
(5, 'Ethan', 1200.00);

---drop TABLE brightlearn.default.orders;

--Question 2:Label each order by its value.
SELECT customer_name,
       amount,
       CASE
       WHEN amount >= 1000 THEN 'High Value'
       WHEN amount BETWEEN 500 and 999.99 THEN 'Medium Value'
       WHEN amount < 500 THEN 'Low Value'
       END AS order_value_category
FROM brightlearn.default.orders;

CREATE TABLE IF NOT EXISTS brightlearn.default.employees1(
    emp_id INT,
    emp_name STRING,
    department STRING,
    salary DECIMAL
);

INSERT INTO brightlearn.default.employees1 VALUES
(1, 'John', 'IT', 85000),
(2, 'Sara', 'HR', 60000),
(3, 'Mark', 'IT', 75000),
(4, 'Lucy', 'Finance', 95000),
(5, 'Tom', 'HR', 55000);

--Question 3: Categorize each employee's position level using both department and salary
SELECT emp_name,
       department,
       salary,
       CASE
       WHEN department = 'IT' AND salary > 80000 THEN 'Senoir IT'
       WHEN department = 'HR' AND salary > 55000 THEN 'Experienced HR'
       ELSE 'Staff'
       END AS position_level
FROM brightlearn.default.employees1;

CREATE TABLE IF NOT EXISTS brightlearn.default.students1(
    student_id INT,
    student_name STRING,
    score INT
);

INSERT INTO brightlearn.default.students1 VALUES
(1, 'Anna', 92),
(2, 'Ben', 76),
(3, 'Cara', 59),
(4, 'Davide', 83),
(5, 'Ella', 68);

--Question 4: Assign each student a letter grade based on their score
SELECT student_name,
        score,
        CASE
        WHEN score >= 90 THEN 'A'
        WHEN score BETWEEN 80 and 89 THEN 'B'
        WHEN score BETWEEN 70 and 79 THEN 'B'
        WHEN score BETWEEN 60 and 69 THEN 'D'
        ELSE 'F'
        END AS grade
FROM brightlearn.default.students1;

DROP TABLE brightlearn.default.students1;
--

CREATE TABLE IF NOT EXISTS brightlearn.default.deliveries(
    delivery_id INT,
    delivery_time_minutes INT
);

INSERT INTO brightlearn.default.deliveries VALUES
( 1, 45),
(2, 80),
(3,  30),
(4, 65),
(5, 100);

--Question 5: Label delivery performance based on the time taken.
SELECT delivery_id,
       delivery_time_minutes,
       CASE
        WHEN delivery_time_minutes <= 30 THEN 'Fast'
        WHEN delivery_time_minutes BETWEEN 31 and 60 THEN 'On Time'
        WHEN delivery_time_minutes > 60 THEN 'Late'
        END AS performance
FROM brightlearn.default.deliveries;

CREATE TABLE IF NOT EXISTS brightlearn.default.tickets
(
    ticket_id INT,
    issue_type STRING,
    priority INT
);

INSERT INTO brightlearn.default.tickets VALUES
(1, 'Login issue', 1),
(2, 'Server down', 3),
(3, 'Slow system', 2),
(4, 'Email error', 2),
(5, 'Password Reset', 1);

--Question 6: Classify tickets based on their priority level.
SELECT ticket_id,
       issue_type,
       priority,
       CASE
        WHEN priority = 3 THEN 'High'
        WHEN priority = 2 THEN 'Medium'
        WHEN priority = 1 THEN 'Low'
        END AS priority_level
FROM brightlearn.default.tickets


--Question 7: Calculate attendance percentage and classify the result
```
